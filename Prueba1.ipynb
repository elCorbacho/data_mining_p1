```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.preprocessing import MinMaxScaler, StandardScaler
from sklearn.decomposition import PCA

print("Bibliotecas importadas correctamente")

print("\n" * 2)
print("==============================")
print("1. CARGA Y EXPLORACION DE DATOS")
print("\n" * 2)


# Cargar el dataset
df = pd.read_csv("student-por.csv")

# Mostrar dimensiones
print("Dimensiones:", df.shape)
print("Filas:", df.shape[0])
print("Columnas:", df.shape[1])

# Mostrar primeras cinco observaciones
display(df.head())

# Mostrar tipos de datos
print(df.dtypes)

# Estadísticas descriptivas
display(df.describe())


print("\n" * 2)
print("==============================")
print("2. DETECCION DE VALORES FALTANTES")
print("\n" * 2)


# Detectar valores nulos por variable
print("Valores faltantes:")
print(df.isnull().sum())

# Detectar duplicados
print("Duplicados:", df.duplicated().sum())

# Crear una copia de trabajo
df_limpio = df.copy()

print("\n" * 2)
print("==============================")
print("3. TRANSFORMACION DE DATOS CATEGORICOS")
print("\n" * 2)

# Codificación binaria del sexo
df_limpio["sex_encoded"] = (
    df_limpio["sex"].map({"F": 0, "M": 1})
)

# Mostrar transformación
display(df_limpio[["sex", "sex_encoded"]].head())

# Distribución por sexo
print(df_limpio["sex"].value_counts())


print("\n" * 2)
print("==============================")
print("4. NORMALIZACION DE DATOS")
print("\n" * 2)

# Seleccionar variables académicas
columnas = ["G1", "G2", "G3"]

# Crear normalizador
scaler = MinMaxScaler()

# Aplicar normalización
df_normalizado = pd.DataFrame(
    scaler.fit_transform(df_limpio[columnas]),
    columns=columnas,
    index=df_limpio.index
)

# Mostrar mínimos y máximos
print("Mínimos:")
print(df_normalizado.min())

print("Máximos:")
print(df_normalizado.max())

display(df_normalizado.head())




print("\n" * 2)
print("==============================")
print("5. DETECCION DE VALORES ATIPICOS")
print("\n" * 2)

plt.figure(figsize=(8, 5))

plt.boxplot(df_limpio["G1"].dropna())

plt.title("Distribución de calificaciones G1")
plt.ylabel("Calificación")
plt.grid(axis="y", alpha=0.3)
plt.show()

Q1 = df_limpio["G1"].quantile(0.25)
Q3 = df_limpio["G1"].quantile(0.75)
IQR = Q3 - Q1

limite_inferior = Q1 - 1.5 * IQR
limite_superior = Q3 + 1.5 * IQR
outliers = df_limpio[
    (df_limpio["G1"] < limite_inferior) |
    (df_limpio["G1"] > limite_superior)
]

print("Límite inferior:", limite_inferior)
print("Límite superior:", limite_superior)
print("Cantidad de outliers:", len(outliers))


print("\n" * 2)
print("==============================")
print("6. INTEGRACION DE DATASETS")
print("\n" * 2)
import pandas as pd
import numpy as np
# Cargar dataset académico original
df = pd.read_csv("student-por.csv")
# Generar identificador según el orden original
df["student_id"] = np.arange(1, len(df) + 1)
# Cargar dataset socioeconómico simulado
df_socio = pd.read_csv(
    "dataset_socioeconomico_simulado.csv"
)
# Integrar mediante LEFT JOIN
df_integrado = pd.merge(
    df,
    df_socio,
    on="student_id",
    how="left",
    validate="one_to_one",
    indicator=True
)
# Comprobar la integridad de los datos
print("Registros originales:", len(df))
print("Registros integrados:", len(df_integrado))
print("Coincidencias:")
print(df_integrado["_merge"].value_counts())
# Revisar resultados
display(df_integrado.head())
# Guardar dataset integrado
df_integrado.drop(columns="_merge").to_csv(
    "estudiantes_integrados.csv",
    index=False
)


print("\n" * 2)
print("==============================")
print("7.1 ESTANDARIZACION DE VARIABLES")
print("\n" * 2)

# Variables para PCA
X = df_limpio[["G1", "G2", "G3"]]
# Estandarización
scaler_pca = StandardScaler()
X_escalado = scaler_pca.fit_transform(X)
# Reducción de tres variables a dos componentes
pca = PCA(n_components=2)
componentes = pca.fit_transform(X_escalado)
df_pca = pd.DataFrame(
    componentes,
    columns=["PC1", "PC2"],
    index=df_limpio.index
)

print("Varianza explicada:")
print(pca.explained_variance_ratio_)
print(
    "Varianza acumulada:",
    pca.explained_variance_ratio_.sum()
)
display(df_pca.head())


print("\n" * 2)
print("==============================")
print("7.2 VISUALIZACION PCA")
print("\n" * 2)
plt.figure(figsize=(9, 6))
plt.scatter(
    df_pca["PC1"],
    df_pca["PC2"],
    alpha=0.6
)
plt.xlabel("Componente principal 1")
plt.ylabel("Componente principal 2")
plt.title("PCA del rendimiento académico")
plt.grid(alpha=0.3)
plt.show()





```
